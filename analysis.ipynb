# Core Metrics

3 core metrics to determine if a real estate market is seller-favored, balanced, or buyer-favored:

1. Months of Inventory: `Active Listings / Monthly Sales = Months of Inventory`

	Calculates how long it would take for all current homes on the market to sell at the current pace of sales if no new homes were listed.

- The Benchmarks:
	- Seller’s Market: Less than 4 months of inventory. Supply is low, and buyers are competing.
	- Balanced Market: 4 to 6 months of inventory. Supply matches demand reasonably well.
	- Buyer’s Market: More than 6 months of inventory. Supply is high, giving buyers the upper hand.


2. Days On Market (DOM) and Sale-to-List Price Ratio

	Must look at how quickly homes move and how much leverage sellers have during negotiations

	- Days on Market (DOM):
		- Seller-Favored: DOM is steadily decreasing or historically low (e.g., under 30 days).
		- Buyer-Favored: DOM is steadily increasing or high (e.g., over 60–90 days).
 	- Sale-to-List Price Ratio:
		-  Seller-Favored: The ratio is close to or above 100%, meaning homes sell at or above asking price.
		- Buyer-Favored: The ratio drops below 95%, showing that sellers are accepting steep discounts.


3. Year-over-Year Velocity

	Real estate is highly seasonal (spring is usually busier than winter). To avoid seasonal confusion, compare the current month's data to the same month in previous years.

	- Seller-Favored Shift: If this October has fewer active listings and higher median prices than last October, the zip code is moving deeper into a seller's market.
	- Buyer-Favored Shift: If active listings are spiking and price growth is flatlining or turning negative compared to the same time last year, the market is shifting toward buyers.

```python
# lets create a function that spits out month_year as 2 columns: month and year 
def split_monthYear(month_year):
    ''' create a function that spits out month_year as 2 columns: month and year ''' 
    string_monthYear = month_year.astype(str)
    year = string_monthYear.str[:4].astype(int)
    month = string_monthYear.str[4:].astype(int)
    return pd.DataFrame({'year': year, 'month': month}, index=month_year.index)

orl_historical[['year', 'month']] = split_monthYear(orl_historical['month_year'])
orl_historical
```
