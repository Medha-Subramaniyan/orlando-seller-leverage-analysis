# Core metrics: seller-favored, balanced, or buyer-favored

## How the two sources work together

| | Redfin | Realtor.com |
|---|---|---|
| Based on | Closed sales | Active listings |
| Timing | Lagging: reflects deals already done | Leading: changes before sales do |
| Period | Rolling 3-month windows | Single months |
| Benchmarks | Industry standards apply | No standard benchmarks |
| Role | Sets the label | Confirms the label, signals direction |

Each zip gets three outputs:
1. **Label:** Seller-favored / Balanced / Buyer-favored (Redfin benchmarks)
2. **Confidence:** High / Mixed (do both sources agree?)
3. **Outlook:** Heating up / Steady / Cooling (leading signals)

## Joining the data

- Key: zip (as string) + Redfin period end date.
- Realtor.com monthly values are averaged over the same 3 months as each
  Redfin window (e.g. Redfin "Aug 2026" = average of Realtor.com Jun, Jul,
  Aug 2026).
- Check after joining: row count, and every one of Jeanine's zips present in
  both sources for each comparison window.

## Dimension 1: Supply

**Primary (Redfin):** `MONTHS OF SUPPLY`
Check: ≈ INVENTORY ÷ (HOMES SOLD ÷ 3). Use INVENTORY, not ACTIVE LISTINGS.

- Seller-favored: under 4 months
- Balanced: 4–6 months
- Buyer-favored: over 6 months
(Benchmark source: [cite]. Redfin's reporting uses 4–5 months as balanced.)

**Supporting (Realtor.com):** `pending_ratio` (pending ÷ active listings)
Should move opposite to months of supply.

**Leading (both):**
- Supply flow: Redfin `NEW LISTINGS ÷ HOMES SOLD`. Above 1 = inventory
  building.
- Realtor.com `new_listing_count` vs. same month last year: a monthly early
  read on new supply.

## Dimension 2: Speed

**Primary (Redfin):** `MEDIAN DAYS ON MARKET` (sold homes)

- Seller-favored: under 30 days
- Balanced: 30–60 days
- Buyer-favored: over 60 days

**Supporting:**
- Redfin `PERCENT OFF MARKET IN TWO WEEKS`: higher = hotter demand.
- Realtor.com `median_days_on_market` (listings): should move in the same
  direction as Redfin's.

**Leading:** Redfin `PENDING SALES` vs. a year earlier. Pending sales lead
closed sales by 1–2 months.

## Dimension 3: Price leverage

**Primary (Redfin):** `AVERAGE SALE TO LIST RATIO`

- Seller-favored: 99% or higher
- Balanced: 95% to under 99%
- Buyer-favored: under 95%
(Balanced-band cutoff is my choice; see decision log.)

**Supporting:** Redfin `SHARE SOLD ABOVE ORIGINAL LIST`. The client-friendly
version ("X% of homes here sold over asking").

**Leading (Realtor.com):** `price_reduced_share`. Sellers cut prices before
they sell, so rising price cuts usually show up before sale-to-list falls.
This is the strongest early-warning signal across both sources.

**Note:** Redfin's sale-to-list compares the sale price to the *final* list
price, after any cuts. A zip can show 98% sale-to-list while 40% of
listings had to cut first. The two together give the full picture:
how often sellers came down (Realtor.com) and how far below the final
price they closed (Redfin).

## Dimension 4: Price trend

- Redfin `MEDIAN SALE PRICE PER SQ.FT.` vs. a year earlier: what buyers
  actually paid.
- Realtor.com `median_listing_price_per_square_foot` vs. a year earlier:
  what sellers are asking.
- Per square foot is used instead of median price, which is distorted when
  the mix of homes changes.
- If asking prices rise while sale prices are flat, sellers are
  overpricing for the market. Useful for listing appointments.

## Turning metrics into outputs

**1. Label (the map)**
Each primary metric (months of supply, days on market, sale-to-list) votes
using its benchmarks. Label = majority vote. If all three disagree,
months of supply decides.

**2. Confidence**
For each dimension, check whether the Realtor.com supporting signal agrees
with the Redfin primary metric (same direction vs. the same window last
year).
- High: all three primary metrics agree, and Realtor.com agrees on at
  least 2 of 3 dimensions.
- Mixed: anything else.

**3. Outlook**
Based on leading signals vs. the same window last year:
- Cooling: price reduced share rising, pending sales falling, supply flow
  above 1. (2 of 3 = Cooling)
- Heating up: the reverse. (2 of 3 = Heating up)
- Steady: otherwise.

Displayed together, e.g.: "Balanced (High confidence), Cooling."

**4. Change since 2022 and 2019 (the bar chart)**
- Headline: change in months of supply, Aug 2022 window → Aug 2026 window.
- Also shown: change in label (e.g. Seller-favored → Balanced).
- 2022 = peak market (distance from the frenzy).
- 2019 = normal market (has the zip normalized or overcorrected?).
- History by source: Realtor.com monthly data and the Redfin Key Metrics
  file cover 2016 onward. If the All Metrics file only starts in 2023,
  2022 months of supply and sale-to-list are reconstructed from the YoY
  columns (approximate); 2019 comparisons use days on market, pending
  sales, new listings, and price reduced share.

## Limitations

- Redfin periods are rolling 3-month windows; labels react to change more
  slowly than single months.
- Redfin sale-to-list measures against the final list price, not the
  original; price cuts before sale are captured separately.
- Low-volume zips are excluded or flagged (see decision log for cutoff).
  Redfin suppresses some values in small zips.
- Reconstructed 2022 values are approximate.
- All property types are combined; condo-heavy zips like 32801 are compared
  against single-family-heavy zips. Property-type split is v2.
- Zips don't match neighborhoods exactly.
- Benchmarks are industry rules of thumb; cutoffs and sources are in the
  decision log.

# Analysis

1. Definition of seller leverage

Seller leverage is the degree of negotiating advantage resale home sellers have over buyers, reflected in how much competition sellers face, how quickly homes sell, and how closely sale prices match asking prices.

Stronger leverage means sellers can generally sell faster, with fewer competing listings and fewer price concessions. It does not require home prices to be rising.

# Methodology Decision Checklist

- [x] **Plain-language definition of seller leverage**
  - Defined as the negotiating advantage resale sellers have over buyers, measured through supply (months of supply), speed (days on market), and price concessions (sale-to-list ratio and price-reduction share). Rising home prices are not required.

- [x] **Reference period for standardizing the score**
  - Use the pooled June 2021–August 2026 study period across all 20 ZIP codes to standardize metrics consistently, while using June–August 2021 as the primary comparison baseline.

- [x] **Metro average: median or mean (from the EDA skew check)**
  - Use the unweighted median across the 20 ZIP codes to reduce outlier influence. For hypothesis tests, compare each ZIP's change against the median of ZIP-level changes.

- [x] **Hypothesis thresholds: how big a gap vs. the metro counts as real for H1, H2, and H3**
  - Use directional differences to evaluate hypotheses and the historical p90 threshold to flag unusually large deviations.
  - **H1:** Evaluate both ZIPs (34787, 32712) across price-reduction share and days on market.
  - **H2:** Compare 34786's leverage-score change against the metro median change.
  - **H3:** Compare the median leverage-score change of affordable ZIPs (34761, 32703) against growth ZIPs (34787, 32712).
  - The p90 threshold is a descriptive benchmark, not proof of statistical significance.

- [x] **Label benchmarks: cite a source for each, and justify the sale-to-list balanced band**
  - Use a three-metric majority vote:
    - **Months of supply:** <4 seller-favored, 4–6 balanced, >6 buyer-favored.
    - **Redfin DOM:** <30 days seller-favored, 30–60 balanced, >60 buyer-favored.
    - **Sale-to-list:** ≥99% seller-favored, 95% to <99% balanced, <95% buyer-favored.
  - Months-of-supply thresholds are informed by industry guidance but require verification against a specific cited source.
  - DOM thresholds are project-specific analytical conventions.
  - The 99% sale-to-list cutoff is calibrated using Orlando's 2019 versus 2021 market data. The 95% floor remains an unvalidated convention.
  - Document the source or calibration rationale for each threshold before finalizing the methodology.

- [x] **Mar–May 2019 as a "normal market" reference: yes or no**
  - **Yes, with an adjustment:** Use June–August 2019 instead to match the season of the 2021 and 2026 comparisons.
  - Treat it as pre-pandemic historical context, not another hypothesis baseline or proof of a universally normal market.

- [x] **Keep confidence and outlook as realtor-facing outputs: yes or no**
  - **Yes to both, but simplify confidence.**
  - **Outlook:** Heating up / steady / cooling, determined using year-over-year leading indicators.
  - **Confidence:** Replace numerical confidence with a Redfin–Realtor.com source-agreement flag (agree / mixed).
  - This communicates uncertainty without complicating the scoring model.

1. Subtract two months from the current observation's date to establish the window's start.

2. Compare dates to identify observations between that start and the current month, inclusive.

3. Aggregate the selected observations for the same ZIP code.

```python
cols_to_avg = realtor.columns.to_list()
cols_to_avg = cols_to_avg[3:]
cols_to_avg
```

```python

realtor['postal_code'] = realtor['postal_code'].astype(str)
realtor['date'] = pd.PeriodIndex(realtor['date'], freq='M')   # works whether date is text or already a period

# 1. Sort observations by ZIP and date
realtor = realtor.sort_values(by=['postal_code', 'date']).reset_index(drop=True)

# 2-3. For each ZIP, for each month: average the window (current month - 2) through current month
rows = []
for zip_code, zip_df in realtor.groupby('postal_code'):
    for current_month in zip_df['date']:
        window_start = current_month - 2
        window = zip_df[(zip_df['date'] >= window_start) & (zip_df['date'] <= current_month)]

        if len(window) < 3:   # incomplete window (a zip's first 2 months, or a missing month): skip it
            continue

        window_avg = window[cols_to_avg].mean().add_suffix('_3m')
        rows.append({'postal_code': zip_code, 'date': current_month, **window_avg})

realtor_3m = pd.DataFrame(rows)

# keep the raw monthly columns next to the 3-month averages
realtor = realtor.merge(realtor_3m, on=['postal_code', 'date'], how='left')
realtor
```

```python
import numpy as np

# check 1: each zip should have exactly 2 empty _3m values (its first 2 months, Jul and Aug 2016)
print(realtor.groupby('postal_code')['price_reduced_share_3m'].apply(lambda s: s.isna().sum()).unique())

# check 2: matches pandas' own rolling(3) (valid because every zip has all its months)
rolling_check = realtor.groupby('postal_code')[cols_to_avg].transform(lambda s: s.rolling(3).mean())
print(all(np.allclose(rolling_check[c], realtor[c + '_3m'], equal_nan=True) for c in cols_to_avg))

# check 3 (by hand): 34786, Jun-Aug 2021 -> the 2021-08 _3m value should equal the average of the 3 raw values
check = realtor[(realtor['postal_code'] == '34786') &
                (realtor['date'].between(pd.Period('2021-06', 'M'), pd.Period('2021-08', 'M')))]
check[['date', 'price_reduced_share', 'price_reduced_share_3m']]
```

### 2) Join Realtor.com to Redfin

- **Key:** zip + `date`. The zip column is `postal_code` in Realtor.com and `REGION NAME` in Redfin, so `merge` uses `left_on` / `right_on`.

- **What to expect for each zip:** Realtor.com runs Jul 2016 to Aug 2026 (122 months). Redfin windows end Sep 2016 to Aug 2026 (120). So each zip should have 120 `both` rows, 2 `left_only` rows (Jul and Aug 2016: Realtor.com months with no Redfin window and no `_3m` value), and 0 `right_only` rows.

```python
redfin['REGION NAME'] = redfin['REGION NAME'].astype(str)   # same key types as realtor
redfin['date'] = pd.PeriodIndex(redfin['date'], freq='M')

joined_outer = realtor.merge(redfin, left_on=['postal_code', 'date'], right_on=['REGION NAME', 'date'],
                             how='outer', indicator=True)
joined_outer['zip'] = joined_outer['postal_code'].fillna(joined_outer['REGION NAME'])

pd.crosstab(joined_outer['zip'], joined_outer['_merge'])
```

## Building the Leverage score

Score = average of 3 dimensions (supply, speed, price leverage), each weighted ⅓. **Higher = more seller leverage.**

| Input | Column | Dimension | Higher value means… | Flip? |
|---|---|---|---|---|
| Months of supply | `rf_months_supply` | supply | less seller leverage | yes |
| DOM (Realtor.com) | `rdc_dom` | speed | less seller leverage | yes |
| DOM (Redfin) | `rf_dom` | speed | less seller leverage | yes |
| Price cut share | `rdc_price_cut_pct` | price leverage | less seller leverage | yes |
| Sale-to-list | `rf_sale_to_list_pct` | price leverage | **more** seller leverage | no |

**Standardizing:** each input becomes a z-score, `(value − reference mean) ÷ reference SD`. The reference is every zip × window inside the study period: windows labeled 2021-08 (Jun–Aug 2021) through 2026-08 (Jun–Aug 2026), so 20 zips × 61 windows. The same fixed reference is used for every window, including 2019, so a metro-wide shift shows up in the score instead of being standardized away.

A score of 0 = the average zip-window in the study period. +1 = one standard deviation more seller leverage than that.

```python
# start from the saved panel (Step 6), so the analysis doesn't depend on the cleaning cells
panel = pd.read_csv('data/processed/orlando_panel_monthly.csv', dtype={'zip': str})
panel['date'] = pd.PeriodIndex(panel['date'], freq='M')

score_inputs = {
    # column:               (dimension, direction)  direction -1 = flip, because higher means LESS seller leverage
    'rf_months_supply':     ('supply', -1),
    'rdc_dom':              ('speed',  -1),
    'rf_dom':               ('speed',  -1),
    'rdc_price_cut_pct':    ('price',  -1),
    'rf_sale_to_list_pct':  ('price',  +1),
}

reference = panel[panel['date'].between(pd.Period('2021-08', 'M'), pd.Period('2026-08', 'M'))]
print(len(reference), 'reference rows =', reference['zip'].nunique(), 'zips x', reference['date'].nunique(), 'windows')

ref_stats = reference[list(score_inputs)].agg(['mean', 'std']).T
ref_stats
```

```python
# standardize against the reference, then flip so higher = more seller leverage
scored = panel.copy()
for col, (dimension, direction) in score_inputs.items():
    scored['z_' + col] = direction * (scored[col] - ref_stats.loc[col, 'mean']) / ref_stats.loc[col, 'std']

z_cols = ['z_' + col for col in score_inputs]

# check: inside the reference period every z should have mean 0 and SD 1
in_ref = scored['date'].between(pd.Period('2021-08', 'M'), pd.Period('2026-08', 'M'))
scored.loc[in_ref, z_cols].agg(['mean', 'std']).round(3)
```

### Average within dimensions, then across them

- **Supply** = z(months of supply)
- **Speed** = average of z(Realtor.com DOM) and z(Redfin DOM)
- **Price leverage** = average of z(price cut share) and z(sale-to-list)
- **`leverage_score`** = average of the 3 dimensions (⅓ each)

**Robustness version (option 2):** `leverage_score_alt` = plain average of 4 inputs, dropping Redfin DOM: months of supply, Realtor.com DOM, price cut share, sale-to-list.

`skipna=False` so a missing input gives a missing score instead of quietly averaging fewer inputs.

```python
scored['dim_supply'] = scored[['z_rf_months_supply']].mean(axis=1, skipna=False)
scored['dim_speed']  = scored[['z_rdc_dom', 'z_rf_dom']].mean(axis=1, skipna=False)
scored['dim_price']  = scored[['z_rdc_price_cut_pct', 'z_rf_sale_to_list_pct']].mean(axis=1, skipna=False)

scored['leverage_score'] = scored[['dim_supply', 'dim_speed', 'dim_price']].mean(axis=1, skipna=False)

# robustness: option 2, four inputs equally weighted (Redfin DOM dropped)
scored['leverage_score_alt'] = scored[['z_rf_months_supply', 'z_rdc_dom',
                                       'z_rdc_price_cut_pct', 'z_rf_sale_to_list_pct']].mean(axis=1, skipna=False)

print('missing scores:', scored[['leverage_score', 'leverage_score_alt']].isna().sum().to_dict())
print('rows scored:', len(scored), '| zips:', scored['zip'].nunique(), '| windows:', scored['date'].nunique())
print('correlation between the two versions:', round(scored['leverage_score'].corr(scored['leverage_score_alt']), 3))
```

# Measure change and test the hypotheses

**How each comparison works** (rules from the Methodology Decision Checklist):

- **Change** = value in the "now" window − value in the baseline window, for each zip.
- **Metro change** = median of the 20 zips' changes (unweighted).
- **Gap** = zip change − metro change. **The sign of the gap decides the direction** (did the zip do better or worse than the metro?).
- **p90 flag:** a gap bigger than the 90th percentile of gaps in a pre-pandemic test period (2017 → 2019, same season, no hypotheses attached) is flagged as *unusually large*. It's descriptive, not proof of statistical significance. The test period covers 2 years while the comparisons cover 4–5, so it probably understates normal drift.

| Comparison | Baseline | Now | Test period for p90 |
|---|---|---|---|
| Primary | Jun–Aug 2021 | Jun–Aug 2026 | Jun–Aug 2017 → Jun–Aug 2019 |
| Robustness | Mar–May 2022 | Mar–May 2026 | Mar–May 2017 → Mar–May 2019 |

Units: score changes are in standard deviations; price cut share in percentage points; DOM in days.

```python
scores = pd.read_csv('data/processed/orlando_scores_monthly.csv', dtype={'zip': str})
scores['date'] = pd.PeriodIndex(scores['date'], freq='M')

comparisons = {
    # name:        (baseline label, now label, test-period start, test-period end)
    'primary':     ('2021-08', '2026-08', '2017-08', '2019-08'),
    'robustness':  ('2022-05', '2026-05', '2017-05', '2019-05'),
}

def change_by_zip(metric, start, end):
    """Each zip's change in `metric` from window `start` to window `end`."""
    by_window = scores.set_index(['date', 'zip'])[metric]
    return by_window.loc[pd.Period(end, 'M')] - by_window.loc[pd.Period(start, 'M')]

def compare_to_metro(metric, comparison):
    """Per zip: baseline, now, change, metro change, gap, and whether the gap is beyond the test-period p90."""
    base, now, test_start, test_end = comparisons[comparison]
    by_window = scores.set_index(['date', 'zip'])[metric]

    change = change_by_zip(metric, base, now)
    test_change = change_by_zip(metric, test_start, test_end)
    p90 = (test_change - test_change.median()).abs().quantile(0.9)

    out = pd.DataFrame({
        'baseline': by_window.loc[pd.Period(base, 'M')],
        'now': by_window.loc[pd.Period(now, 'M')],
        'change': change,
    })
    out['metro_change'] = change.median()
    out['gap'] = out['change'] - out['metro_change']
    out['p90'] = p90
    out['beyond_p90'] = out['gap'].abs() > p90
    out.insert(0, 'comparison', comparison)
    out.insert(0, 'metric', metric)
    return out

# p90 thresholds for every metric used below
metrics = ['leverage_score', 'leverage_score_alt', 'rdc_price_cut_pct', 'rdc_dom', 'rf_dom']
pd.DataFrame({c: {m: compare_to_metro(m, c)['p90'].iloc[0] for m in metrics} for c in comparisons}).round(2)
```

### Score change for every zip

Negative change = the zip lost seller leverage. Positive gap = it lost **less** than the metro (or gained more).

```python
score_changes = pd.concat([compare_to_metro(m, c)
                           for m in ['leverage_score', 'leverage_score_alt']
                           for c in comparisons])

# one row per zip: change and gap for each comparison x score version
score_summary = score_changes.reset_index().pivot(index='zip', columns=['metric', 'comparison'], values=['change', 'gap'])
score_summary = score_summary.sort_values(('gap', 'leverage_score', 'primary'))
print('metro change:', score_changes.groupby(['metric', 'comparison'])['metro_change'].first().round(2).to_dict())
score_summary.round(2)
```

```python
# primary comparison, main score: each zip's gap vs. the metro (focus zips highlighted)
primary = compare_to_metro('leverage_score', 'primary').sort_values('gap')
colors = ['tab:orange' if z in focus_zips else 'tab:gray' for z in primary.index]

ax = primary['gap'].plot.barh(color=colors, figsize=(8, 7),
                              title='Leverage score change vs. metro, Jun–Aug 2021 → Jun–Aug 2026')
ax.axvline(0, color='black', linewidth=0.8)
for edge in (-primary['p90'].iloc[0], primary['p90'].iloc[0]):
    ax.axvline(edge, color='grey', linestyle='--', linewidth=0.8)
ax.set_xlabel('gap (SD): + = lost less leverage than the metro; dashed = test-period p90')
```

### H1: New construction hurt resale sellers most

*In 34787 and 32712, price cut share and days on market will have risen more than the metro average. Rejected if both zips' increases are at or below the metro average.*

H1 uses the raw metrics, not the score. Days on market is checked in both sources. A **positive gap** supports H1 (rose more than the metro).

```python
h1_zips = ['34787', '32712']
h1_metrics = ['rdc_price_cut_pct', 'rdc_dom', 'rf_dom']

h1 = pd.concat([compare_to_metro(m, c).loc[h1_zips] for m in h1_metrics for c in comparisons])
h1['rose_more_than_metro'] = h1['gap'] > 0
h1.round(2)
```

```python
# the stated rule, per metric and comparison: rejected only if BOTH zips are at or below the metro
h1_rule = (h1.reset_index()
             .groupby(['metric', 'comparison'])
             .agg(zips_above_metro=('rose_more_than_metro', 'sum'),
                  zips_beyond_p90=('beyond_p90', 'sum')))
h1_rule['rejected_by_stated_rule'] = h1_rule['zips_above_metro'] == 0
h1_rule['both_zips_above'] = h1_rule['zips_above_metro'] == 2
h1_rule
```

### H3: Affordable zips lost less than growth zips

*34761 and 32703 will have lost less leverage than 34787 and 32712. Rejected if their drop is the same as or larger than the growth zips'.*

Compares the **median** score change of each group. A **positive difference** (affordable − growth) supports H3. There's no p90 for a group difference, so the table also shows each zip's own gap vs. the metro.

```python
groups = {'affordable': ['34761', '32703'], 'growth': ['34787', '32712']}

rows = []
for metric in ['leverage_score', 'leverage_score_alt']:
    for comparison in comparisons:
        result = compare_to_metro(metric, comparison)
        affordable = result.loc[groups['affordable'], 'change'].median()
        growth = result.loc[groups['growth'], 'change'].median()
        rows.append({'metric': metric, 'comparison': comparison,
                     'affordable_median_change': affordable,
                     'growth_median_change': growth,
                     'difference': affordable - growth,
                     'affordable_lost_less': affordable > growth})
h3 = pd.DataFrame(rows)
h3.round(2)
```

```python
# each H3 zip's own result, primary comparison
compare_to_metro('leverage_score', 'primary').loc[groups['affordable'] + groups['growth'],
                                                  ['change', 'metro_change', 'gap', 'beyond_p90']].round(2)
```

### Do any conclusions flip?

# Realtor Facing Outputs

Four outputs for each zip, built for Jeanine rather than for the hypothesis tests:

1. **Label**: Seller-favored / Balanced / Buyer-favored, in the baseline window (Jun–Aug 2021) and now (Jun–Aug 2026)
2. **Confidence**: do Realtor.com and Redfin agree on the direction over the past year? (agree / mixed)
3. **Outlook**: Heating up / Steady / Cooling, from leading signals vs. a year earlier
4. **Price trend**: asking $/sq ft vs. sale $/sq ft over the past year

"A year earlier" = the same window one year before (Jun–Aug 2025 for the current window), so seasonality cancels out.

```python
panel = pd.read_csv('data/processed/orlando_panel_monthly.csv', dtype={'zip': str})
panel['date'] = pd.PeriodIndex(panel['date'], freq='M')
by_window = panel.set_index(['date', 'zip'])

NOW, YEAR_AGO, BASE = pd.Period('2026-08', 'M'), pd.Period('2025-08', 'M'), pd.Period('2021-08', 'M')
now, year_ago, base = by_window.loc[NOW], by_window.loc[YEAR_AGO], by_window.loc[BASE]
yoy = now.select_dtypes('number') - year_ago.select_dtypes('number')   # change vs. a year earlier
```

### 1) Label: majority vote of three benchmarks

| Metric | Seller-favored | Balanced | Buyer-favored |
|---|---|---|---|
| Months of supply | under 4 | 4–6 | over 6 |
| Redfin DOM | under 30 days | 30–60 days | over 60 days |
| Sale-to-list | 99% or higher | 95% to under 99% | under 95% |

Label = the label at least 2 of 3 metrics give. If all three disagree, months of supply decides. These use Redfin's published columns.

```python
def vote_supply(mos):
    return 'Seller-favored' if mos < 4 else ('Balanced' if mos <= 6 else 'Buyer-favored')

def vote_speed(dom):
    return 'Seller-favored' if dom < 30 else ('Balanced' if dom <= 60 else 'Buyer-favored')

def vote_price(sale_to_list):
    return 'Seller-favored' if sale_to_list >= 99 else ('Balanced' if sale_to_list >= 95 else 'Buyer-favored')

def label_window(window):
    """Votes and the majority label for every zip in one window."""
    votes = pd.DataFrame({
        'vote_supply': window['rf_months_supply'].map(vote_supply),
        'vote_speed':  window['rf_dom'].map(vote_speed),
        'vote_price':  window['rf_sale_to_list_pct'].map(vote_price),
    })
    def majority(row):
        counts = row.value_counts()
        return counts.index[0] if counts.iloc[0] >= 2 else row['vote_supply']   # 3-way split: supply decides
    votes['label'] = votes.apply(majority, axis=1)
    return votes

labels_base = label_window(base)
labels_now = label_window(now)

labels = labels_base[['label']].rename(columns={'label': 'label_2021'}).join(labels_now.add_suffix('_2026'))
labels = labels.rename(columns={'label_2026': 'label_now'})
print(pd.crosstab(labels['label_2021'], labels['label_now']), '\n')
labels
```

### 2) Confidence: do the two sources agree?

For each dimension, compare the direction of the Redfin metric with its Realtor.com partner over the past year:

| Dimension | Redfin | Realtor.com partner | They agree when… |
|---|---|---|---|
| Supply | months of supply | pending ratio | they move in **opposite** directions (more supply ↔ fewer pendings per listing) |
| Speed | DOM (sold homes) | DOM (listings) | they move in the **same** direction |
| Price | sale-to-list | price cut share | they move in **opposite** directions (closing below list ↔ more price cuts) |

**agree** = the sources agree on at least 2 of 3 dimensions; otherwise **mixed**.

```python
def same_direction(a, b):
    return np.sign(a) == np.sign(b)

agreement = pd.DataFrame({
    'supply_agrees': ~same_direction(yoy['rf_months_supply'], yoy['rdc_pending_ratio']),   # opposite
    'speed_agrees':   same_direction(yoy['rf_dom'], yoy['rdc_dom']),                         # same
    'price_agrees':  ~same_direction(yoy['rf_sale_to_list_pct'], yoy['rdc_price_cut_pct']),  # opposite
})
agreement['dimensions_agreeing'] = agreement.sum(axis=1)
agreement['confidence'] = np.where(agreement['dimensions_agreeing'] >= 2, 'agree', 'mixed')
print(agreement['confidence'].value_counts().to_dict())
agreement
```

### 3) Outlook: leading signals vs. a year earlier

| Signal | Cooling when… | Heating up when… |
|---|---|---|
| Price cut share (Realtor.com) | rising | falling |
| Pending sales (Redfin) | falling | rising |
| Supply flow = new listings ÷ homes sold (Redfin) | rising | falling |

**Cooling** if at least 2 of 3 signals say cooling; **Heating up** if at least 2 say heating; otherwise **Steady**.

Why supply flow uses the *change* and not "above 1": in Orlando, new listings exceed homes sold in 70–100% of zips every year, even at the 2021 peak, so "above 1" would call nearly every zip Cooling.

```python
supply_flow_now = now['rf_new_listings'] / now['rf_homes_sold']
supply_flow_year_ago = year_ago['rf_new_listings'] / year_ago['rf_homes_sold']

signals = pd.DataFrame({
    'price_cuts_change':    yoy['rdc_price_cut_pct'],             # + = cooling
    'pending_sales_change': yoy['rf_pending_sales'],              # - = cooling
    'supply_flow_change':   supply_flow_now - supply_flow_year_ago,  # + = cooling
})
cooling = (signals['price_cuts_change'] > 0).astype(int) + (signals['pending_sales_change'] < 0) + (signals['supply_flow_change'] > 0)
heating = (signals['price_cuts_change'] < 0).astype(int) + (signals['pending_sales_change'] > 0) + (signals['supply_flow_change'] < 0)

signals['cooling_signals'] = cooling
signals['heating_signals'] = heating
signals['outlook'] = np.select([cooling >= 2, heating >= 2], ['Cooling', 'Heating up'], default='Steady')
print(signals['outlook'].value_counts().to_dict())
signals.round(2)
```

### 4) Price trend: asking vs. sale $/sq ft over the past year

- **Asking** = Realtor.com median listing price per sq ft; **sale** = Redfin median sale price per sq ft. Both are % change vs. a year earlier.
- **Flag** when asking prices are rising (more than +2%) while sale prices are flat or falling (+2% or less). Sellers may be pricing ahead of what buyers pay, which is useful in listing appointments.
- "Flat **or falling**" rather than only flat: a zip where asking prices rise while sale prices drop (e.g. +7% vs. −6%) is the strongest version of the mismatch, and a flat-only rule would miss it.
- The 2% cutoffs are a judgment call, set in `RISING` and `FLAT` below. Record whatever you choose in the decision log.

```python
RISING = 2.0   # % per year
FLAT = 2.0     # % per year: sale prices at or below this count as flat or falling

price_trend = pd.DataFrame({
    'asking_ppsf_yoy_pct': (now['rdc_list_ppsf'] / year_ago['rdc_list_ppsf'] - 1) * 100,
    'sale_ppsf_yoy_pct':   (now['rf_sale_ppsf'] / year_ago['rf_sale_ppsf'] - 1) * 100,
})
price_trend['asking_minus_sale_pts'] = price_trend['asking_ppsf_yoy_pct'] - price_trend['sale_ppsf_yoy_pct']
price_trend['asking_outpacing_sales'] = (price_trend['asking_ppsf_yoy_pct'] > RISING) & (price_trend['sale_ppsf_yoy_pct'] <= FLAT)
print('flagged zips:', price_trend.index[price_trend['asking_outpacing_sales']].tolist())
price_trend.sort_values('asking_minus_sale_pts', ascending=False).round(1)
```

### All four outputs, one row per zip

Displayed together, e.g. "Balanced (sources agree), Cooling". Focus zips first.

```python
realtor_outputs = (labels[['label_2021', 'label_now']]
                   .join(agreement['confidence'])
                   .join(signals['outlook'])
                   .join(price_trend[['asking_ppsf_yoy_pct', 'sale_ppsf_yoy_pct', 'asking_outpacing_sales']].round(1)))
realtor_outputs.insert(0, 'is_focus', realtor_outputs.index.isin(focus_zips))
realtor_outputs['summary'] = (realtor_outputs['label_now'] + ' (sources ' + realtor_outputs['confidence'] + '), '
                              + realtor_outputs['outlook'])
realtor_outputs = realtor_outputs.sort_values(['is_focus', 'label_now'], ascending=[False, True])

realtor_outputs.to_csv('data/processed/realtor_outputs.csv')
realtor_outputs
```

### Check 3: low-volume or fragile zips

- **Low volume:** fewer than 50 homes sold in a 3-month window (about 17 a month) in any comparison window. Medians and shares from that few sales move a lot by chance. The 50 cutoff is a judgment call.
- **Fragile:** shift each window by one month either way (baseline Jul / Aug / Sep 2021 windows × now Jun / Jul / Aug 2026 windows = 9 combinations). If the zip's direction vs. the metro isn't the same in all 9, one window choice could decide the result. Same test for the robustness comparison.

```python
MIN_HOMES_SOLD = 50   # per 3-month window

key_windows = [pd.Period(m, 'M') for m in ['2021-08', '2026-08', '2022-05', '2026-05']]
homes_sold = scores.set_index(['date', 'zip'])['rf_homes_sold']
min_sold = pd.concat([homes_sold.loc[w] for w in key_windows], axis=1).min(axis=1)

def direction_stability(base_labels, now_labels):
    """Share of baseline x now window combinations in which each zip lost less than the metro."""
    lost_less = []
    for b in base_labels:
        for n in now_labels:
            change = score_by_window.loc[pd.Period(n, 'M')] - score_by_window.loc[pd.Period(b, 'M')]
            lost_less.append(change > change.median())
    return pd.concat(lost_less, axis=1).mean(axis=1)

stability = pd.DataFrame({
    'min_homes_sold_3mo': min_sold,
    'primary_share_lost_less': direction_stability(['2021-07', '2021-08', '2021-09'], ['2026-06', '2026-07', '2026-08']),
    'robust_share_lost_less': direction_stability(['2022-04', '2022-05', '2022-06'], ['2026-04', '2026-05', '2026-06']),
})
stability['low_volume'] = stability['min_homes_sold_3mo'] < MIN_HOMES_SOLD
stability['fragile_primary'] = ~stability['primary_share_lost_less'].isin([0, 1])
stability['fragile_robust'] = ~stability['robust_share_lost_less'].isin([0, 1])
stability['is_focus'] = stability.index.isin(focus_zips)

print('low volume:', stability.index[stability['low_volume']].tolist())
print('fragile (primary):', stability.index[stability['fragile_primary']].tolist())
print('fragile (robustness):', stability.index[stability['fragile_robust']].tolist())
stability.sort_values(['is_focus', 'min_homes_sold_3mo'], ascending=[False, True]).round(2)
```

### Talking points for Jeanine's zips

One line per focus zip: today's label, how its loss of leverage compares with Orlando overall, two concrete numbers clients understand, and a confidence note. Generated from the tables above. Edit the wording before using it with clients.

- "About as much as Orlando overall" = the gap is within the p90 line; "noticeably better/worse" = beyond it.
- Confidence notes: sources mixed, result depends on the window chosen (fragile), or low sales volume.

```python
window_values = scores.set_index(['date', 'zip'])
then, latest = window_values.loc[pd.Period('2021-08', 'M')], window_values.loc[pd.Period('2026-08', 'M')]

def talking_point(zip_code):
    row, result, check = realtor_outputs.loc[zip_code], primary.loc[zip_code], stability.loc[zip_code]
    city = scores.loc[scores['zip'] == zip_code, 'zip_name'].iloc[0].split(',')[0].title()

    if not result['beyond_p90']:
        comparison = 'lost about as much seller leverage as Orlando overall'
    elif result['gap'] > 0:
        comparison = 'held up noticeably better than Orlando overall'
    else:
        comparison = 'lost noticeably more seller leverage than Orlando overall'

    notes = []
    if row['confidence'] == 'mixed':
        notes.append('Realtor.com and Redfin disagree on the past year')
    if check['fragile_primary'] or check['fragile_robust']:
        notes.append('the comparison depends on which months you pick')
    if check['low_volume']:
        notes.append('few sales, so numbers move a lot')
    confidence = 'Caveat: ' + '; '.join(notes) + '.' if notes else 'Both sources agree, and the result holds across comparisons.'

    return (f"{zip_code} ({city}): {row['label_now']} today, down from {row['label_2021']} in summer 2021; it {comparison}. "
            f"Price cuts went from {then.loc[zip_code, 'rdc_price_cut_pct']:.0f}% to {latest.loc[zip_code, 'rdc_price_cut_pct']:.0f}% of listings, "
            f"and homes now take {latest.loc[zip_code, 'rf_dom']:.0f} days to sell (vs. {then.loc[zip_code, 'rf_dom']:.0f}). "
            f"Past year: {row['outlook'].lower()}. {confidence}")

talking_points = pd.Series({z: talking_point(z) for z in focus_zips}, name='talking_point')
for point in talking_points:
    print(point, '\n')
talking_points.to_csv('data/processed/talking_points.csv')
```
