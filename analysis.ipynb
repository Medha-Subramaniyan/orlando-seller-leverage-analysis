# Research Question
Where in Orlando have sellers lost leverage since 2023, and where do they still have it?

# Hypothesis

**H1: New construction hurt resale sellers most.**  
		In 34787 and 32712, price cut share and days on market will have risen more than the metro average since 2022, because builders' incentives pull buyers away from resale homes.  
		_Rejected if:_ both zips' increases are at or below the metro average.


**H2: Windermere held up.**  
		In 34786, the leverage score will have dropped less than the metro average since 2022, because scarce lakefront and large-lot homes and less rate-sensitive buyers protect sellers.  
		_Rejected if:_ 34786's drop is at or above the metro average. If that happens, the Horizon West portion of the zip may be dragging it down.


**H3: Affordable zips**
 34761 and 32703 will have lost less leverage than 34787 and 32712, because they have less new-construction competition. _Rejected if:_ their drop is the same as or larger than the growth zips'.

### Filtering national zip level data down to Orlando, Florida

```python
# the only way we can filter to Orlando, Florida is by spltting zip_name into city, state

#
orl = august.copy()
orl[['city', 'state']] = orl['zip_name'].str.split(',', expand=True)
orl['state'] = orl['state'].str.strip()
orl['city'] = orl['city'].str.strip()


```

```python
orl = orl[orl['state'] == 'fl']
orl['state'].nunique() #we've filtered down to florida! (only 1 unique)
orl['city'].nunique() #interesting, there are 418 municipalities in florida
```
