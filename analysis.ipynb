# Core metrics: seller-favored, balanced, or buyer-favored

## How the two sources work together

| | Redfin | Realtor.com |
|---|---|---|
| Based on | Closed sales | Active listings |
| Timing | Lagging: reflects deals already done | Leading: changes before sales do |
| Period | Rolling 3-month windows | Single months |
| Benchmarks | Industry standards apply | No standard benchmarks |
| Role | Sets the label | Confirms the label, signals direction |

Each zip gets three outputs:
1. **Label:** Seller-favored / Balanced / Buyer-favored (Redfin benchmarks)
2. **Confidence:** High / Mixed (do both sources agree?)
3. **Outlook:** Heating up / Steady / Cooling (leading signals)

## Joining the data

- Key: zip (as string) + Redfin period end date.
- Realtor.com monthly values are averaged over the same 3 months as each
  Redfin window (e.g. Redfin "Aug 2026" = average of Realtor.com Jun, Jul,
  Aug 2026).
- Check after joining: row count, and every one of Jeanine's zips present in
  both sources for each comparison window.

## Dimension 1: Supply

**Primary (Redfin):** `MONTHS OF SUPPLY`
Check: ≈ INVENTORY ÷ (HOMES SOLD ÷ 3). Use INVENTORY, not ACTIVE LISTINGS.

- Seller-favored: under 4 months
- Balanced: 4–6 months
- Buyer-favored: over 6 months
(Benchmark source: [cite]. Redfin's reporting uses 4–5 months as balanced.)

**Supporting (Realtor.com):** `pending_ratio` (pending ÷ active listings)
Should move opposite to months of supply.

**Leading (both):**
- Supply flow: Redfin `NEW LISTINGS ÷ HOMES SOLD`. Above 1 = inventory
  building.
- Realtor.com `new_listing_count` vs. same month last year: a monthly early
  read on new supply.

## Dimension 2: Speed

**Primary (Redfin):** `MEDIAN DAYS ON MARKET` (sold homes)

- Seller-favored: under 30 days
- Balanced: 30–60 days
- Buyer-favored: over 60 days

**Supporting:**
- Redfin `PERCENT OFF MARKET IN TWO WEEKS`: higher = hotter demand.
- Realtor.com `median_days_on_market` (listings): should move in the same
  direction as Redfin's.

**Leading:** Redfin `PENDING SALES` vs. a year earlier. Pending sales lead
closed sales by 1–2 months.

## Dimension 3: Price leverage

**Primary (Redfin):** `AVERAGE SALE TO LIST RATIO`

- Seller-favored: 99% or higher
- Balanced: 95% to under 99%
- Buyer-favored: under 95%
(Balanced-band cutoff is my choice; see decision log.)

**Supporting:** Redfin `SHARE SOLD ABOVE ORIGINAL LIST`. The client-friendly
version ("X% of homes here sold over asking").

**Leading (Realtor.com):** `price_reduced_share`. Sellers cut prices before
they sell, so rising price cuts usually show up before sale-to-list falls.
This is the strongest early-warning signal across both sources.

**Note:** Redfin's sale-to-list compares the sale price to the *final* list
price, after any cuts. A zip can show 98% sale-to-list while 40% of
listings had to cut first. The two together give the full picture:
how often sellers came down (Realtor.com) and how far below the final
price they closed (Redfin).

## Dimension 4: Price trend

- Redfin `MEDIAN SALE PRICE PER SQ.FT.` vs. a year earlier: what buyers
  actually paid.
- Realtor.com `median_listing_price_per_square_foot` vs. a year earlier:
  what sellers are asking.
- Per square foot is used instead of median price, which is distorted when
  the mix of homes changes.
- If asking prices rise while sale prices are flat, sellers are
  overpricing for the market. Useful for listing appointments.

## Turning metrics into outputs

**1. Label (the map)**
Each primary metric (months of supply, days on market, sale-to-list) votes
using its benchmarks. Label = majority vote. If all three disagree,
months of supply decides.

**2. Confidence**
For each dimension, check whether the Realtor.com supporting signal agrees
with the Redfin primary metric (same direction vs. the same window last
year).
- High: all three primary metrics agree, and Realtor.com agrees on at
  least 2 of 3 dimensions.
- Mixed: anything else.

**3. Outlook**
Based on leading signals vs. the same window last year:
- Cooling: price reduced share rising, pending sales falling, supply flow
  above 1. (2 of 3 = Cooling)
- Heating up: the reverse. (2 of 3 = Heating up)
- Steady: otherwise.

Displayed together, e.g.: "Balanced (High confidence), Cooling."

**4. Change since 2022 and 2019 (the bar chart)**
- Headline: change in months of supply, Aug 2022 window → Aug 2026 window.
- Also shown: change in label (e.g. Seller-favored → Balanced).
- 2022 = peak market (distance from the frenzy).
- 2019 = normal market (has the zip normalized or overcorrected?).
- History by source: Realtor.com monthly data and the Redfin Key Metrics
  file cover 2016 onward. If the All Metrics file only starts in 2023,
  2022 months of supply and sale-to-list are reconstructed from the YoY
  columns (approximate); 2019 comparisons use days on market, pending
  sales, new listings, and price reduced share.

## Limitations

- Redfin periods are rolling 3-month windows; labels react to change more
  slowly than single months.
- Redfin sale-to-list measures against the final list price, not the
  original; price cuts before sale are captured separately.
- Low-volume zips are excluded or flagged (see decision log for cutoff).
  Redfin suppresses some values in small zips.
- Reconstructed 2022 values are approximate.
- All property types are combined; condo-heavy zips like 32801 are compared
  against single-family-heavy zips. Property-type split is v2.
- Zips don't match neighborhoods exactly.
- Benchmarks are industry rules of thumb; cutoffs and sources are in the
  decision log.

# Analysis

1. Definition of seller leverage

Seller leverage is the degree of negotiating advantage resale home sellers have over buyers, reflected in how much competition sellers face, how quickly homes sell, and how closely sale prices match asking prices.

Stronger leverage means sellers can generally sell faster, with fewer competing listings and fewer price concessions. It does not require home prices to be rising.

# Methodology Decision Checklist

- [x] **Plain-language definition of seller leverage**
  - Defined as the negotiating advantage resale sellers have over buyers, measured through supply (months of supply), speed (days on market), and price concessions (sale-to-list ratio and price-reduction share). Rising home prices are not required.

- [x] **Reference period for standardizing the score**
  - Use the pooled June 2021–August 2026 study period across all 20 ZIP codes to standardize metrics consistently, while using June–August 2021 as the primary comparison baseline.

- [x] **Metro average: median or mean (from the EDA skew check)**
  - Use the unweighted median across the 20 ZIP codes to reduce outlier influence. For hypothesis tests, compare each ZIP's change against the median of ZIP-level changes.

- [x] **Hypothesis thresholds: how big a gap vs. the metro counts as real for H1, H2, and H3**
  - Use directional differences to evaluate hypotheses and the historical p90 threshold to flag unusually large deviations.
  - **H1:** Evaluate both ZIPs (34787, 32712) across price-reduction share and days on market.
  - **H2:** Compare 34786's leverage-score change against the metro median change.
  - **H3:** Compare the median leverage-score change of affordable ZIPs (34761, 32703) against growth ZIPs (34787, 32712).
  - The p90 threshold is a descriptive benchmark, not proof of statistical significance.

- [x] **Label benchmarks: cite a source for each, and justify the sale-to-list balanced band**
  - Use a three-metric majority vote:
    - **Months of supply:** <4 seller-favored, 4–6 balanced, >6 buyer-favored.
    - **Redfin DOM:** <30 days seller-favored, 30–60 balanced, >60 buyer-favored.
    - **Sale-to-list:** ≥99% seller-favored, 95% to <99% balanced, <95% buyer-favored.
  - Months-of-supply thresholds are informed by industry guidance but require verification against a specific cited source.
  - DOM thresholds are project-specific analytical conventions.
  - The 99% sale-to-list cutoff is calibrated using Orlando's 2019 versus 2021 market data. The 95% floor remains an unvalidated convention.
  - Document the source or calibration rationale for each threshold before finalizing the methodology.

- [x] **Mar–May 2019 as a "normal market" reference: yes or no**
  - **Yes, with an adjustment:** Use June–August 2019 instead to match the season of the 2021 and 2026 comparisons.
  - Treat it as pre-pandemic historical context, not another hypothesis baseline or proof of a universally normal market.

- [x] **Keep confidence and outlook as realtor-facing outputs: yes or no**
  - **Yes to both, but simplify confidence.**
  - **Outlook:** Heating up / steady / cooling, determined using year-over-year leading indicators.
  - **Confidence:** Replace numerical confidence with a Redfin–Realtor.com source-agreement flag (agree / mixed).
  - This communicates uncertainty without complicating the scoring model.

### 1) Avg Realtor.com over 3 months per zip

FOR EACH ZIP
1. Subtract two months from the current observation's date to establish the window's start.

2. Compare dates to identify observations between that start and the current month, inclusive.

3. Aggregate the selected observations for the same ZIP code.

```python
cols_to_avg = realtor.columns.to_list()
cols_to_avg = cols_to_avg[3:]
cols_to_avg
```

```python

realtor['postal_code'] = realtor['postal_code'].astype(str)
realtor['date'] = pd.PeriodIndex(realtor['date'], freq='M')   # works whether date is text or already a period

# 1. Sort observations by ZIP and date
realtor = realtor.sort_values(by=['postal_code', 'date']).reset_index(drop=True)

# 2-3. For each ZIP, for each month: average the window (current month - 2) through current month
rows = []
for zip_code, zip_df in realtor.groupby('postal_code'):
    for current_month in zip_df['date']:
        window_start = current_month - 2
        window = zip_df[(zip_df['date'] >= window_start) & (zip_df['date'] <= current_month)]

        if len(window) < 3:   # incomplete window (a zip's first 2 months, or a missing month): skip it
            continue

        window_avg = window[cols_to_avg].mean().add_suffix('_3m')
        rows.append({'postal_code': zip_code, 'date': current_month, **window_avg})

realtor_3m = pd.DataFrame(rows)

# keep the raw monthly columns next to the 3-month averages
realtor = realtor.merge(realtor_3m, on=['postal_code', 'date'], how='left')
realtor
```

```python
import numpy as np

# check 1: each zip should have exactly 2 empty _3m values (its first 2 months, Jul and Aug 2016)
print(realtor.groupby('postal_code')['price_reduced_share_3m'].apply(lambda s: s.isna().sum()).unique())

# check 2: matches pandas' own rolling(3) (valid because every zip has all its months)
rolling_check = realtor.groupby('postal_code')[cols_to_avg].transform(lambda s: s.rolling(3).mean())
print(all(np.allclose(rolling_check[c], realtor[c + '_3m'], equal_nan=True) for c in cols_to_avg))

# check 3 (by hand): 34786, Jun-Aug 2021 -> the 2021-08 _3m value should equal the average of the 3 raw values
check = realtor[(realtor['postal_code'] == '34786') &
                (realtor['date'].between(pd.Period('2021-06', 'M'), pd.Period('2021-08', 'M')))]
check[['date', 'price_reduced_share', 'price_reduced_share_3m']]
```

Note: `new_listing_count_3m` is a **monthly average**. Redfin's `NEW LISTINGS` and `HOMES SOLD` are **3-month totals**, about 3× larger. Compare direction across sources, not levels.

### 2) Join Realtor.com to Redfin

- **Key:** zip + `date`. The zip column is `postal_code` in Realtor.com and `REGION NAME` in Redfin, so `merge` uses `left_on` / `right_on`.
- **Outer join first**, with `indicator=True`, so rows that exist in only one source show up instead of disappearing.
- **What to expect for each zip:** Realtor.com runs Jul 2016 to Aug 2026 (122 months). Redfin windows end Sep 2016 to Aug 2026 (120). So each zip should have 120 `both` rows, 2 `left_only` rows (Jul and Aug 2016: Realtor.com months with no Redfin window and no `_3m` value), and 0 `right_only` rows.
- The only shared column name is `date`, which is part of the key, so no `_x` / `_y` duplicates.

```python
redfin['REGION NAME'] = redfin['REGION NAME'].astype(str)   # same key types as realtor
redfin['date'] = pd.PeriodIndex(redfin['date'], freq='M')

joined_outer = realtor.merge(redfin, left_on=['postal_code', 'date'], right_on=['REGION NAME', 'date'],
                             how='outer', indicator=True)
joined_outer['zip'] = joined_outer['postal_code'].fillna(joined_outer['REGION NAME'])

pd.crosstab(joined_outer['zip'], joined_outer['_merge'])
```

```python
# which months are only in one source?
print('left_only months: ', sorted(joined_outer.loc[joined_outer['_merge'] == 'left_only', 'date'].astype(str).unique()))
print('right_only months:', sorted(joined_outer.loc[joined_outer['_merge'] == 'right_only', 'date'].astype(str).unique()))

# anything not 'both' inside the study period means something is wrong
study_period = joined_outer['date'].between(pd.Period('2021-06', 'M'), pd.Period('2026-08', 'M'))
print('non-both rows in study period:', (study_period & (joined_outer['_merge'] != 'both')).sum())
```

**Decision:** keep only the `both` rows. The 40 `left_only` rows are Jul and Aug 2016, which have no Redfin window and no 3-month average, and they fall outside the study period. Nothing is `right_only`, and every study-period row matched.

```python
joined = joined_outer[joined_outer['_merge'] == 'both'].drop(columns=['_merge', 'zip']).reset_index(drop=True)

print(joined.shape)                                   # expect 2400 rows = 20 zips x 120 windows
print(joined['postal_code'].nunique(), joined['date'].nunique())
print(joined.filter(like='_3m').isna().sum().sum())   # expect 0
```

### 3) Check months of supply

Redfin's `MONTHS OF SUPPLY` should be ≈ `INVENTORY ÷ (HOMES SOLD ÷ 3)`. `HOMES SOLD` is a 3-month total, so ÷ 3 gives monthly sales.

**Tolerance, decided before looking:** Redfin reports months of supply to 1 decimal, so differences within ±0.1 are rounding. A gap regularly above ±0.5 would mean the column means something different from what we think.

```python
joined['mos_check'] = joined['INVENTORY'] / (joined['HOMES SOLD'] / 3)
mos_diff = joined['mos_check'] - joined['MONTHS OF SUPPLY']

print(mos_diff.describe())
print('share within ±0.1:', round((mos_diff.abs() <= 0.1).mean(), 3))
print('share within ±0.5:', round((mos_diff.abs() <= 0.5).mean(), 3))
mos_diff.hist(bins=40)
```

```python
# the rows outside rounding: which zips and months, and are they low-sales zips?
joined.loc[mos_diff.abs() > 0.1, ['postal_code', 'date', 'INVENTORY', 'HOMES SOLD', 'MONTHS OF SUPPLY', 'mos_check']]
```

**Result (from the data when this was written):** the formula is confirmed, with a small, consistent difference.
- 89% of rows are within ±0.1 and **all** are within ±0.5. The largest gap is 0.25 months.
- The calculated value is slightly **lower** than Redfin's (median −0.04). Redfin probably measures inventory or sales slightly differently, for example inventory at a different point in the window.
- Because the difference is small and consistent, it doesn't affect changes between windows. For labels, use Redfin's own column, so the 4- and 6-month cutoffs are applied to the published number.

### 4) Tidy the columns

- **Prefixes:** `rdc_` = Realtor.com (3-month average), `rf_` = Redfin (3-month window). Keep only the columns the analysis uses.
- **Scale:** Realtor.com's price cut share is a fraction (0.22). Redfin's shares are percentages (97.5). **Decision: everything in percent.** The label benchmarks (99%, 95%) and client talking points ("22% of listings cut") are in percent. `rdc_pending_ratio` stays a ratio, because it's pending ÷ active and can go above 1.
- `rdc_new_listings_mo_avg` is a monthly average. `rf_new_listings` and `rf_homes_sold` are 3-month totals.
- Raw single-month Realtor.com values stay in `orlando_realtor.csv`, for measuring noise on single months.

```python
rename_map = {
    'postal_code': 'zip',
    # Realtor.com, 3-month averages
    'price_reduced_share_3m': 'rdc_price_cut_pct',
    'median_days_on_market_3m': 'rdc_dom',
    'pending_ratio_3m': 'rdc_pending_ratio',
    'new_listing_count_3m': 'rdc_new_listings_mo_avg',
    'median_listing_price_per_square_foot_3m': 'rdc_list_ppsf',
    # Redfin, 3-month windows
    'MONTHS OF SUPPLY': 'rf_months_supply',
    'MEDIAN DAYS ON MARKET (DAYS)': 'rf_dom',
    'AVERAGE SALE TO LIST RATIO (%)': 'rf_sale_to_list_pct',
    'SHARE SOLD ABOVE ORIGINAL LIST (%)': 'rf_sold_above_list_pct',
    'PERCENT OFF MARKET IN TWO WEEKS (%)': 'rf_off_market_2wk_pct',
    'PENDING SALES': 'rf_pending_sales',
    'NEW LISTINGS': 'rf_new_listings',
    'HOMES SOLD': 'rf_homes_sold',
    'INVENTORY': 'rf_inventory',
    'MEDIAN SALE PRICE PER SQ.FT. ($)': 'rf_sale_ppsf',
}

panel = joined[['date', 'zip_name'] + list(rename_map)].rename(columns=rename_map)
panel['rdc_price_cut_pct'] = panel['rdc_price_cut_pct'] * 100   # fraction -> percent, to match Redfin
panel['is_focus'] = panel['zip'].isin(focus_zips)                # Jeanine's zips

metric_cols = [c for c in panel.columns if c.startswith(('rdc_', 'rf_'))]
panel = panel[['zip', 'zip_name', 'is_focus', 'date'] + metric_cols]
panel
```

### 5) Pull out the comparison windows

Every comparison uses 3-month windows, labeled by their **last** month:

| Role | Window | `date` label |
|---|---|---|
| Primary baseline | Jun–Aug 2021 | 2021-08 |
| Primary now | Jun–Aug 2026 | 2026-08 |
| Robustness baseline | Mar–May 2022 | 2022-05 |
| Robustness now | Mar–May 2026 | 2026-05 |
| Context (pre-pandemic) | Jun–Aug 2019 | 2019-08 |

The window labels live in one dictionary, so later code uses names like `'primary_base'` instead of typed-out dates. Each window should have 20 rows (one per zip) and no nulls in the Tier 1 columns.

```python
windows = {
    'primary_base': '2021-08',   # Jun–Aug 2021
    'primary_now':  '2026-08',   # Jun–Aug 2026
    'robust_base':  '2022-05',   # Mar–May 2022
    'robust_now':   '2026-05',   # Mar–May 2026
    'context_2019': '2019-08',   # Jun–Aug 2019
}
label_to_window = {pd.Period(month, 'M'): name for name, month in windows.items()}

window_panel = panel[panel['date'].isin(label_to_window)].copy()
window_panel.insert(0, 'window', window_panel['date'].map(label_to_window))

tier1 = ['rdc_price_cut_pct', 'rdc_dom', 'rf_dom', 'rf_months_supply', 'rf_sale_to_list_pct']
print(window_panel['window'].value_counts())        # expect 20 each
print(window_panel[tier1].isna().sum())             # expect 0 each

# metro median per window: a quick look before any scoring
window_panel.groupby('window')[tier1].median().round(2)
```
